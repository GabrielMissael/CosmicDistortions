# Python

Python es un lenguaje de programación interpretado (no compilado). Un lenguaje de programación interpretado es uno donde se "traduce" en tiempo real tu código a lenguaje de maquina, contratio a uno compilado, donde primero se traduce todo el código a lenguaje maquina, y finalmente se envía.

Es un lenguaje:
  * Fácil de aprender
  * De tipado dinámico (no necesitamos declarar el tipo de variables)
  * Se intenta que se "lea como el inglés", es decir, que sea muy fácil de leer aunque no sepas qué hace el código.
  * Usa indentaciones en lugar de llaves {} para marcar inicios y cierres de bloques de código
